```python
import pandas as pd

icu = pd.read_csv("icustays.csv")

print(icu.shape)
print(icu.columns.tolist())
print(icu.head())
```

```python
print(icu["los"].describe())
```

```python
icu["first_careunit"].value_counts()
```

```python
icu.groupby("first_careunit")["los"] \
   .mean() \
   .sort_values(ascending=False)
```

```python
icu.nlargest(
    10,
    "los"
)[
    [
        "first_careunit",
        "last_careunit",
        "los"
    ]
]
```

```python
icu.to_csv("icustays_processed.csv", index=False)
```
