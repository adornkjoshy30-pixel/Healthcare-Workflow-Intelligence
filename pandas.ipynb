```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
print("Libraries laoded sucessfully")
```

```python
#Section 1
# goal - create a small student marks dataset
student_data = {
    "Name" : ["Asha", "Rahul", "Meera", "Arjun", "Sara", "Dev"],
    "Subject" : ["Maths", "Science", "Math", "Science", "Math", "Science"],
    "Marks": [85, 72, 91, 44, 67, 58],
    "Study_Hours": [5, 3,6,1,4,2]
}
```

```python
student_data
```

```python
df = pd.DataFrame(student_data)
df.head()
```

```python
print("First five rows")
df.head()
```

```python
print("\nDataset shape (rows, columsn):")
print(df.shape)
```

```python
print('\nColumn name:')
print(df.columns.tolist())
```

```python
print("\nONly Name and Marks:")
print(df[["Name", "Marks"]])
```

```python
high_scores = df[df["Marks"] >= 70]
print("\nStudents scoring 70 or above:")
print(high_scores)
```

```python
sorted_df = df.sort_values("Marks", ascending=False)
print(sorted_df)
```

```python
df["Result"] = np.where(df["Marks"] >= 50, "Pass", "Need support")
print("\nAdded result column:")
print(df)
```

```python
average_by_subject =df.groupby("Subject") ["Marks"].mean()
average_by_subject
```
