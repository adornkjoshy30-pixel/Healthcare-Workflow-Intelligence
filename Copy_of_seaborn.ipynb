```python
#basic setup of seaborn
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Load a sample dataset
df=sns.load_dataset("iris")
df.head()
```

```python
#Distribution plots
sns.histplot(df['sepal_length'])
```

```python
sns.kdeplot(df['sepal_length'])
```

```python
sns.histplot(df['sepal_length'],kde=True)
```

```python
sns.displot(df,x='sepal_length',kde=True)
```

```python
sns.rugplot(df['sepal_length'])
```

```python
sns.histplot(df['sepal_length'],kde=True)
```

```python
sns.rugplot(df['sepal_length'],color='red')
```

```python
sns.ecdfplot(df['sepal_length'])
```

```python
#Categorical plots
sns.countplot(x='species',data=df)
```

```python
sns.barplot(x='species',y='sepal_length',data=df)
```

```python
sns.boxplot(x='species',y='sepal_length',data=df)
```

```python
sns.violinplot(x='species',y='sepal_length',data=df)
```

```python
sns.stripplot(x='species',y='sepal_length',data=df)
```

```python
sns.swarmplot(x='species',y='sepal_length',data=df)
```

```python
sns.scatterplot(x='sepal_length',y='sepal_width',data=df)
```

```python
sns.lineplot(x='sepal_length',y='sepal_width',data=df)
```

```python
sns.relplot(x='sepal_length',y='sepal_width',data=df)
```

```python
#Matrix/Heatmaps
corr=df.corr(numeric_only=True)
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
sns.clustermap(corr,cmap='coolwarm')
```

```python
sns.pairplot(df,hue='species')
```

```python
sns.jointplot(x='sepal_length',y='sepal_width',data=df)
```

```python
sns.FacetGrid(df,col='species').map(sns.histplot,'sepal_length')
```

```python
#Regression plots
sns.regplot(x='sepal_length',y='sepal_width',data=df)
```

```python
sns.lmplot(x='sepal_length',y='sepal_width',data=df)
```
