```python

import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error

def main():
    # 1. Load the sample housing dataset
    data = {
        'House_ID': [1, 2, 3, 4, 5, 6],
        'Length (m)': [15, 20, 12, 25, 18, 22],
        'Width (m)': [10, 15, 8, 20, 12, 18],
        'Rooms': [3, 4, 2, 5, 3, 4],
        'Age (Years)': [5, 2, 20, 1, 10, 8],
        'Price (Units)': [250000, 450000, 180000, 700000, 310000, 520000]
    }

    df = pd.DataFrame(data)

    # Define features (X) and target variable (y) - excluding House_ID
    X = df[['Length (m)', 'Width (m)', 'Rooms', 'Age (Years)']]
    y = df['Price (Units)']

    # 2. Split input data into training and testing subsets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

    # 3. Initialize and train the Random Forest Regressor with specified trees and feature subsets
    rf_model = RandomForestRegressor(
        n_estimators=50,
        max_features='sqrt',
        bootstrap=True,
        random_state=42
    )
    rf_model.fit(X_train, y_train)

    # 4. Evaluate performance (Console Output)
    y_pred = rf_model.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)

    print(f"R-squared (R2): {r2:.4f}")
    print(f"Mean Squared Error (MSE): {mse:.4f}")

    # 5. Export predictions for the test set ("predictions.csv")
    pred_df = X_test.copy()
    pred_df['Actual_Price'] = y_test
    pred_df['Predicted_Price'] = y_pred
    pred_df.to_csv("predictions.csv", index=False)
    print("Predictions saved to 'predictions.csv'")

    # 6. Calculate and save feature importance ranking ("feature_importance.txt")
    importances = rf_model.feature_importances_
    importance_df = pd.DataFrame({
        'Feature': X.columns,
        'Importance': importances
    }).sort_values(by='Importance', ascending=False)

    with open("feature_importance.txt", "w") as f:
        f.write("Feature Importance Ranking (Most to Least Important):\n")
        f.write("===================================================\n")
        for idx, row in importance_df.iterrows():
            f.write(f"{row['Feature']}: {row['Importance']:.4f}\n")

    print("Feature importance ranking saved to 'feature_importance.txt'")

if __name__ == "__main__":
    main()
```
