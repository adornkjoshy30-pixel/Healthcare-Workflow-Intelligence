```python
import random
import gradio as gr
```

```python
def get_scores(move1, move2):
    if move1 == "Cooperate" and move2 == "Cooperate":
        return 3, 3
    elif move1 == "Betray" and move2 == "Cooperate":
        return 5, 0
    elif move1 == "Cooperate" and move2 == "Betray":
        return 0, 5
    else:
        return 1, 1

def ai_move(strategy, opponent_history=None):
    if strategy == "Always Cooperate":
        return "Cooperate"
    elif strategy == "Always Betray":
        return "Betray"
    elif strategy == "Random":
        return random.choice(["Cooperate", "Betray"])
    elif strategy == "Tit-for-Tat":
        if not opponent_history:
            return "Cooperate"
        return opponent_history[-1]
    return "Cooperate"
```

```python
def play_game(player1_strategy, player2_strategy, rounds):
    score1 = 0
    score2 = 0
    history1 = []
    history2 = []
    result = ""

    for i in range(rounds):

        move1 = ai_move(player1_strategy, history2)
        move2 = ai_move(player2_strategy, history1)


        history1.append(move1)
        history2.append(move2)

        points1, points2 = get_scores(move1, move2)
        score1 += points1
        score2 += points2

        result += f"--- Round {i+1} ---\n"
        result += f"Player A: {move1} | Player B: {move2}\n"
        result += f"Scores this round: {points1}, {points2}\n\n"

    result += "====================\n"
    result += "   FINAL SCORES\n"
    result += "====================\n"
    result += f"Player A Total: {score1}\n"
    result += f"Player B Total: {score2}\n"

    if score1 > score2:
        result += "\nWINNER: Player A 🏆"
    elif score2 > score1:
        result += "\nWINNER: Player B 🏆"
    else:
        result += "\nRESULT: It's a Tie! 🤝"

    return result
```
